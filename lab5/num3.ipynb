## Практическое задание

В файле water.txt представлено 61 наблюдение. Каждое наблюдение – город в
Англии и Уэльсе. Города дополнительно поделены на северные и южные. Для
каждого города известны средняя годовая смертность на 100000 населения (по
данным 1958–1964) и концентрация кальция в питьевой воде (в частях на
миллион). Чем выше концентрация кальция, тем жёстче вода. 

```python
import pandas as pd

df = pd.read_csv('14water.txt', sep = '\t')
print(df.head())
```

```python
south_df = df.query("location == 'Sout'")
south_df.describe()[['mortalit', 'hardness']]
```

```python
north_df = df.query("location == 'Nort'")
north_df.describe()[['mortalit', 'hardness']]
```

Южные города характеризуются более низкой средней смертностью и заметно более высокой средней жёсткостью воды. Северные города, напротив, демонстрируют повышенный уровень смертности и более мягкую воду.

2) Постройте 95% доверительные интервалы для средней годовой смертности по
всем южным и северным городам. Отличаются ли границы интервалов?

```python
from statsmodels.stats.weightstats import _tconfint_generic
import numpy as np

def mean_ci(series):
    n = series.shape[0]
    m = series.mean()
    se = series.std(ddof=1) / np.sqrt(n)
    return _tconfint_generic(m, se, n - 1, alpha=0.05/2, alternative='two-sided')

ci_south_mort = mean_ci(south_df['mortalit'])
ci_north_mort = mean_ci(north_df['mortalit'])

print( "95% interval south:", ci_south_mort)
print( "95% interval north:", ci_north_mort)
```

Интервалы практически не пересекаются: диапазон смертности на Севере полностью смещён вверх относительно Южных городов. Это означает статистически значимое отличие средних уровней смертности.

Постройте 95% доверительные интервалы для средней концентрации кальция в
питьевой воде для южных и северных городов. Отличаются ли границы
интервалов?

```python
ci_south_hard = mean_ci(south_df['hardness'])
ci_north_hard = mean_ci(north_df['hardness'])


print( "95% interval south:", ci_south_hard)
print( "95% interval north:", ci_north_hard)
```

Интервальные оценки концентрации кальция в воде существенно различаются. Южные города имеют значительно более высокую жёсткость воды, различие статистически значимо.

Анализ данных показывает две ключевые закономерности:
1) Смертность в северных городах стабильно выше, что подтверждается как описательными статистиками, так и доверительными интервалами.
2) Концентрация кальция в питьевой воде значительно выше в южных городах.
