 Есть ли в количественном признаке выбросы?
Какое значение является выбросом и почему?

Для поиска выбросов используется метод межквартильного размаха (IQR). Признак memSize имеет сильно неоднородное распределение с различиями на несколько порядков (от 0.00003 GB до 288 GB), что делает IQR неэффективным. Чтобы исправить это, данные преобразуются в логарифмический масштаб, уменьшая влияние различий порядков. После логарифмирования вычисляются квартильные значения, межквартильный размах и устанавливаются границы нормальных значений. Значения вне этих границ считаются выбросами.


```python
import numpy as np

# Логарифмирование признака (игнорируем нули и NaN)
df['memSize_log'] = np.log10(df['memSize'])

# Расчёт квартилей
q75, q25 = np.percentile(df['memSize_log'].dropna(), [75, 25])
iqr = q75 - q25

print("IQR:", iqr)

# Границы нормальных значений
lower_bound = q25 - 1.5 * iqr
upper_bound = q75 + 1.5 * iqr

# Поиск выбросов
outliers = df[(df['memSize_log'] < lower_bound) | (df['memSize_log'] > upper_bound)]
outliers
```

После логарифмического преобразования признака memSize и применения межквартильного размаха (IQR) были выявлены две группы выбросов:

Очень большие значения памяти (288 GB) — современные дата-центр GPU.

Очень малые значения памяти (0.002–0.004 GB) — встроенные и мобильные видеочипы конца 1990-х – начала 2000-х годов.

Использование IQR без логарифмирования было невозможно, поскольку исходные данные имели многопорядковую природу и метод не выделял выбросы.
