Обучите модель линейной регрессии на данных согласно вашему варианту

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score


X = np.random.uniform(200, 600, 200)[:, None]

def f(x):
    return 2 * x + np.random.random() * 4


y = np.array([f(x) for x in X])
```

```python
plt.scatter(X, y)
plt.title("X vs y")
plt.show()
```

```python
# Модель линейной регрессии
lr = LinearRegression()
lr.fit(X, y)

# Линия регрессии
X_ = np.arange(250, 600, 10)[:, np.newaxis]
y_lr = lr.predict(X_)

plt.scatter(X, y)
plt.plot(X_, y_lr)
plt.title("Linear Regression Fit")
plt.show()
```

Выведите
угловой коэффициент, точку пересечения с осью ординат и точность.

```python
# Угловой коэффициент (slope)
slope = lr.coef_[0]
print("Угловой коэффициент (slope):", slope)
```

```python
# Точка пересечения с осью ординат (intercept)
intercept = lr.intercept_
print("Пересечение с осью ординат (intercept):", intercept)
```

```python
# Метрики: R² и MSE
y_pred_lr = lr.predict(X)
mse_lr = mean_squared_error(y, y_pred_lr)
r2_lr = r2_score(y, y_pred_lr)

print(f"R-квадрат: {r2_lr}")
print(f"MSE: {mse_lr}")
```
